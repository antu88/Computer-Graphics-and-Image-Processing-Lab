```python
import matplotlib.pyplot as plt
r = int(input("Enter radius: "))
x = 0
y = r
p = 3 - 2 * r

points = []

print("\ni\tX\tY\tP")
print("------------------------")

i = 0

while x <= y:

    print(i, "\t", x, "\t", y, "\t", p)

    # 8 symmetric points
    points += [
        (x, y), (-x, y),
        (x, -y), (-x, -y),
        (y, x), (-y, x),
        (y, -x), (-y, -x)
    ]

    if p < 0:
        p = p + 4*x + 6
        x = x + 1

    else:
        p = p + 4*(x-y) + 10
        x = x + 1
        y = y - 1

    i = i + 1

X = [point[0] for point in points]
Y = [point[1] for point in points]

plt.scatter(X, Y)

plt.grid()
plt.axis("equal")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Bresenham's Circle")

plt.show()
```

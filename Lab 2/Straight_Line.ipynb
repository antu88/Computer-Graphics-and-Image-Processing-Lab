```python
import matplotlib.pyplot as plt
```

```python
x1 = int(input("Enter x1: "))
y1 = int(input("Enter y1: "))

x2 = int(input("Enter x2: "))
y2 = int(input("Enter y2: "))

m = (y2 - y1) / (x2 - x1)

b = y1 - m * x1

print("\nSlope (m) =", m)
print("Intercept (b) =", b)

x_points = []
y_points = []

if abs(m) <= 1:

    print("\nUsing Equation: y = mx + b")
    print("Since |m| <= 1, increment X\n")

    step = 1 if x2 >= x1 else -1

    print("--------------------------------")
    print("     x        y = mx + b")
    print("--------------------------------")

    x = x1

    while True:

        y = m * x + b

        x_points.append(x)
        y_points.append(y)

        if y.is_integer():
            y_display = int(y)
        else:
            y_display = round(y, 2)

        print(f"{x:^10} {y_display:^15}")

        if x == x2:
            break

        x += step

else:

    print("\nUsing Equation: x = (y - b) / m")
    print("Since |m| > 1, increment Y\n")

    step = 1 if y2 >= y1 else -1

    print("--------------------------------")
    print("     y        x = (y-b)/m")
    print("--------------------------------")

    y = y1

    while True:

        x = (y - b) / m

        x_points.append(x)
        y_points.append(y)

        if x.is_integer():
            x_display = int(x)
        else:
            x_display = round(x, 2)

        print(f"{y:^10} {x_display:^15}")

        if y == y2:
            break

        y += step
```

```python
plt.figure(figsize=(8, 6))

plt.plot(
    x_points,
    y_points,
    'o-',
    label='Scan Converted Points'
)

plt.xlabel("X")
plt.ylabel("Y")

plt.title("Scan Conversion of Straight Line")

plt.grid(True)
plt.legend()

plt.axis("equal")

plt.show()
```
